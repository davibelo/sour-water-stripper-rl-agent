```python
import pandas as pd
import numpy as np
```

```python
DATA_ID = "04"
FIGURES_FOLDER = 'figures'
```

```python
RESULTS_FILE = f"sim_results-{DATA_ID}.csv"
CLEANED_FILE = f"sim_results_cleaned-{DATA_ID}.csv"
df = pd.read_csv(RESULTS_FILE)
df.head(5)
```

```python
print('Data rows:', df.shape[0])
```

```python
# Observing data tail
df.tail(5)
```

```python
# Observing a sample
SAMPLE_SIZE = 10
start = np.random.randint(0, len(df) - SAMPLE_SIZE + 1)
sample_df = df.iloc[start : start + SAMPLE_SIZE]

sample_df.head(SAMPLE_SIZE)
```

```python
df.describe()
```

```python
# Check for NaN values
nan_check = df.isna().sum()
print("NaN values per column:\n", nan_check)
```

```python
# Check for zero values
zero_check = (df == 0).sum()
print("\nZero values per column:\n", zero_check)
```

```python
# Filter rows with NaN values
nan_rows = df[df.isna().any(axis=1)]
print("Rows with NaN values:\n", nan_rows)
```

```python
# Drop rows with any NaN values
df = df.dropna()

# Optionally, reset the index if needed
df = df.reset_index(drop=True)

# Check for NaN values
nan_check = df.isna().sum()
print("NaN values per column:\n", nan_check)
```

```python
df.shape
```

```python
# Export processed csv
df.to_csv(CLEANED_FILE, index=False)
```

```python
# Print differences between original and cleaned data
df1 = pd.read_csv(RESULTS_FILE)
df2 = pd.read_csv(CLEANED_FILE)

diff_rows = df1[df1.ne(df2).any(axis=1)]
print(diff_rows)
```
